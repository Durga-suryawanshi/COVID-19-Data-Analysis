```python
summary = {
    'Total Confirmed Cases': df['Daily Confirmed'].sum(),
    'Total Recovered': df['Daily Recovered'].sum(),
    'Total Deaths': df['Daily Deceased'].sum(),
    'Average Daily Cases': df['Daily Confirmed'].mean(),
    'Average Daily Recoveries': df['Daily Recovered'].mean(),
    'Average Daily Deaths': df['Daily Deceased'].mean()
}

summary
```

```python
peak_data = {
    'Peak Cases': df['Daily Confirmed'].max(),
    'Peak Cases Date': df.loc[df['Daily Confirmed'].idxmax(), 'Date_YMD'],
    'Peak Recoveries': df['Daily Recovered'].max(),
    'Peak Recoveries Date': df.loc[df['Daily Recovered'].idxmax(), 'Date_YMD'],
    'Peak Deaths': df['Daily Deceased'].max(),
    'Peak Deaths Date': df.loc[df['Daily Deceased'].idxmax(), 'Date_YMD']
}

peak_data
```

```python
plt.figure(figsize=(14, 7))

plt.plot(df['Date_YMD'], df['Daily Confirmed'], label='Daily Confirmed')
plt.plot(df['Date_YMD'], df['Daily Recovered'], label='Daily Recovered')
plt.plot(df['Date_YMD'], df['Daily Deceased'], label='Daily Deceased')

plt.title('COVID-19 Daily Cases, Recoveries and Deaths')
plt.xlabel('Date')
plt.ylabel('Number of People')

plt.legend()
plt.xticks(rotation=45)
plt.tight_layout()

plt.show()
```

```python
peak_cases = df.loc[df['Daily Confirmed'].idxmax()]

print("Peak Cases:", peak_cases['Daily Confirmed'])
print("Peak Cases Date:", peak_cases['Date_YMD'])
```

```python
peak_recovery = df.loc[df['Daily Recovered'].idxmax()]

print("Peak Recoveries:", peak_recovery['Daily Recovered'])
print("Peak Recoveries Date:", peak_recovery['Date_YMD'])
```

```python
peak_deaths = df.loc[df['Daily Deceased'].idxmax()]

print("Peak Deaths:", peak_deaths['Daily Deceased'])
print("Peak Deaths Date:", peak_deaths['Date_YMD'])
```
